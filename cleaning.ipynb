```python
outlets['Review_Count'] = outlets['Review_Count'].astype(str).str.replace(",", "", regex= False).astype(int)
```

```python
socioeconomic.info()
```

```python
socioeconomic['Region_Area'] = socioeconomic['Region_Area'].str.replace(",", "", regex= False).astype(float)
socioeconomic['Disposable Income per capita'] = socioeconomic['Disposable Income per capita'].str.replace(",", "", regex= False).astype(int)
socioeconomic['Total Tourism Flow'] = socioeconomic['Total Tourism Flow'].str.replace(',','',regex=False).astype(float)
socioeconomic['Foreign Tourism Flow'] = socioeconomic['Foreign Tourism Flow'].str.replace(',','',regex=False).astype(float)
socioeconomic['Median Income'] = socioeconomic['Median Income'].str.replace(',','',regex= False).astype(int)
```

```python
socioeconomic.info()
```

```python
duplicate_coordinate_rows = outlets[
    outlets.duplicated(
        subset=["Latitude", "Longitude"],
        keep=False
    )
].sort_values(["Latitude", "Longitude"])

#To display all the results in a scrolldown option
from IPython.display import display, HTML

html_table = duplicate_coordinate_rows.loc[:, ['Outlet ID', 'Brand', 'City', 'Customer_Rating','Review_Count', 'Latitude', 'Longitude', 'Address']].to_html(index=False)

display(HTML(f"""
<div style="
    max-height: 300px;
    overflow-y: auto;
    overflow-x: auto;
    border: 1px solid #555;
    padding: 8px;
">
{html_table}
</div>
"""))
```
