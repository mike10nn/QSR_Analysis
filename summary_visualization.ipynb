```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
outlets = pd.read_csv("Data/QSR_outlet_dataset_cleaned.csv", keep_default_na= False)
socioeconomic = pd.read_csv("Data/QSR_socioeconomic_dataset_cleaned.csv")
```

```python
# Bar chart of outlets by QSR brand

outlets["Brand"].value_counts().plot(kind="bar", figsize=(10, 4))

plt.title("Number of QSR Outlets by Brand")
plt.xlabel("Brand")
plt.ylabel("Number of Outlets")
plt.xticks(rotation=45, ha="right")
plt.show()
```

This bar chart confirms that the dataset is strongly concentrated among a few major brands. McDonald's has the largest number of outlets, followed by La Piadineria, Autogrill, and Old Wild West. Several brands have much smaller representation.


```python
# Bar chart of outlets by region

outlets["Region"].value_counts().plot(kind="bar", figsize=(10, 4))

plt.title("Number of QSR Outlets by Region")
plt.xlabel("Region")
plt.ylabel("Number of Outlets")
plt.xticks(rotation=45, ha="right")
plt.show()
```

This bar chart confirms that the QSR outlets are strongly concentrated among a few regions. Lombardia has the largest number of outlets, while Molise has the lowest.


```python
# Bar chart of outlets by location type

outlets["Location_Type"].value_counts().plot(kind="bar", figsize=(10, 3))

plt.title("Number of QSR Outlets by Location Type")
plt.xlabel("Location Type")
plt.ylabel("Number of Outlets")
plt.xticks(rotation=45, ha="right")
plt.show()
```

Commercial areas and shopping malls dominate, suggesting that QSR outlets are mainly located in high-traffic consumer spaces.


```python
# Bar chart of outlets by cleaned price range

outlets["Price_Range_Cleaned"].value_counts().plot(kind="bar", figsize=(8, 4))

plt.title("Number of QSR Outlets by Price Range")
plt.xlabel("Price Range")
plt.ylabel("Number of Outlets")
plt.xticks(rotation=0)
plt.show()
```

The results with the price range show that most are in the lower and middle price ranges. That tells us most of the outlets are affordable QSRs.

```python
# Histogram of customer ratings

outlets["Customer_Rating"].plot(kind="hist", bins=20, figsize=(8, 4))

plt.title("Distribution of Customer Ratings")
plt.xlabel("Customer Rating")
plt.ylabel("Number of Outlets")
plt.show()
```

The histogram shows how customer ratings are distributed across QSR outlets. 
Most ratings appear to be concentrated around the middle-to-high range, especially near 4 stars. 
This suggests that QSR outlets generally receive moderate to positive customer ratings, although there are still some lower-rated outlets.


```python
# Histogram of review counts

outlets["Review_Count"].plot(kind="hist", bins=30, figsize=(8, 4))

plt.title("Distribution of Review Counts")
plt.xlabel("Review Count")
plt.ylabel("Number of Outlets")
plt.show()
```

The review count histogram shows that review counts are highly uneven across outlets. Most outlets have relatively low or moderate numbers of reviews, while a smaller number of outlets have very high review counts. This suggests that the review-count variable is skewed.


```python
# Histogram of review counts using log scale

outlets["Review_Count"].plot(kind="hist", bins=30, figsize=(8, 4), logx=True)

plt.title("Distribution of Review Counts (Log Scale)")
plt.xlabel("Review Count")
plt.ylabel("Number of Outlets")
plt.show()
```

Here i performed a log scale on the review counts data. The log-scale histogram makes the review count distribution easier to interpret. 


```python
# Histogram of regional population

(socioeconomic["Population"]/1000000).plot(kind="hist", bins=20, figsize=(8, 4))

plt.title("Distribution of Regional Population")
plt.xlabel("Population in millions")
plt.ylabel("Number of Regions")
plt.show()
```

This histogram shows that regional population is unevenly distributed. Most regions have relatively smaller populations, while a few regions have much larger populations. This matters because regions with larger populations may naturally have more QSR outlets, so later in the analysis i'll consider outlet density relative to population, not only raw outlet counts.


```python
# Histogram of regional population density

socioeconomic["Population Density"].plot(kind="hist", bins=20, figsize=(8, 4))

plt.title("Distribution of Regional Population Density")
plt.xlabel("Population Density (people per km^2)")
plt.ylabel("Number of Regions")
plt.show()
```

This histogram shows how many regions fall into different population-density ranges. Most regions are in lower or middle density ranges, while a few regions are much more densely populated. This matters because QSR outlets may be more common in dense regions where there is more foot traffic, urban activity, and commercial concentration.


```python
# Histogram of total tourism flow in millions

(socioeconomic["Total Tourism Flow"] / 1000000).plot(kind="hist", bins=20, figsize=(8, 4))

plt.title("Distribution of Total Tourism Flow")
plt.xlabel("Total Tourism Flow (millions)")
plt.ylabel("Number of Regions")
plt.show()
```

This histogram shows that tourism flow is unevenly distributed across regions. Most regions have relatively lower tourism flow, while a few regions receive very high numbers of visitors. This matters because tourism may increase demand for QSR outlets, especially in regions with major cities, transport hubs, and tourist destinations.


```python
# Histogram of disposable income per capita

socioeconomic["Disposable Income per capita"].plot(kind="hist", bins=10, figsize=(8, 4))

plt.title("Distribution of Disposable Income per Capita")
plt.xlabel("Disposable Income per Capita")
plt.ylabel("Number of Regions")
plt.show()
```

This histogram shows how disposable income per capita varies across regions. The income distribution helps describe the economic context of each region. This may be important later when comparing QSR outlet concentration, brand presence, or price range across richer and poorer regions.


```python
# Histogram of median income

socioeconomic["Median Income"].plot(kind="hist", bins=10, figsize=(8, 4))

plt.title("Distribution of Median Income")
plt.xlabel("Median Income")
plt.ylabel("Number of Regions")
plt.show()
```

This histogram shows how median income varies across regions. Median income provides another measure of regional economic conditions and is less affected by extreme values than an average. This may be useful later when comparing QSR outlet concentration, brand presence, or price range across regions with different income levels.
