```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
outlets = pd.read_csv("Data/QSR_outlet_dataset_cleaned.csv", keep_default_na= False)
```

```python
# Bar chart of outlets by QSR brand

outlets["Brand"].value_counts().plot(kind="bar", figsize=(10, 4))

plt.title("Number of QSR Outlets by Brand")
plt.xlabel("Brand")
plt.ylabel("Number of Outlets")
plt.xticks(rotation=45, ha="right")
plt.show()
```

This bar chart confirms that the dataset is strongly concentrated among a few major brands. McDonald's has the largest number of outlets, followed by La Piadineria, Autogrill, and Old Wild West. Several brands have much smaller representation.


```python
# Bar chart of outlets by region

outlets["Region"].value_counts().plot(kind="bar", figsize=(10, 4))

plt.title("Number of QSR Outlets by Region")
plt.xlabel("Region")
plt.ylabel("Number of Outlets")
plt.xticks(rotation=45, ha="right")
plt.show()
```

This bar chart confirms that the QSR outlets are strongly concentrated among a few regions. Lombardia has the largest number of outlets, while Molise has the lowest.
