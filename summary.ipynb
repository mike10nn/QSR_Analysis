```python
# Average customer rating and review count by brand

outlets.groupby("Brand")[["Customer_Rating", "Review_Count"]].mean().sort_values(
    by="Customer_Rating",
    ascending=False
)
```

This table shows the average customer rating and average review count for each QSR brand. It helps compare how brands perform in terms of customer satisfaction and customer attention.

However, these averages should be interpreted carefully because the number of outlets is not the same for every brand. Brands with many outlets, such as McDonald's or La Piadineria, have more stable averages because they are based on many observations. Brands with only a few outlets can have averages that are strongly influenced by one or two locations.

Therefore, this table is useful for identifying initial patterns, but it shoul d not be treated as a final ranking of brand performance without also considering the number of outlets for each brand.


```python
# Brand-level summary with outlet count, average rating, and average review count

outlets.groupby("Brand").agg(
    Outlet_Count=("Outlet ID", "count"),
    Average_Rating=("Customer_Rating", "mean"),
    Average_Review_Count=("Review_Count", "mean")
).sort_values(by="Average_Rating", ascending=False)
```

This table adds the number of outlets for each brand alongside the average rating and average review count. This makes the comparison more reliable because we can see whether each brand average is based on many outlets or only a few. Brands with larger outlet counts provide more stable averages, while brands with small outlet counts should be interpreted carefully.


```python
# Region-level summary with outlet count, average rating, and average review count

outlets.groupby("Region").agg(
    Outlet_Count=("Outlet ID", "count"),
    Average_Rating=("Customer_Rating", "mean"),
    Average_Review_Count=("Review_Count", "mean")
).sort_values(by="Outlet_Count", ascending=False)
```

This table summarizes QSR outlets by region. It shows where outlets are concentrated and compares average customer rating and average review count across regions.

Regions with many outlets, such as Lombardia and Lazio, have more stable averages because they are based on many observations. Regions with very few outlets, such as Molise or Valle d'Aosta, should be interpreted more carefully because a small number of outlets can strongly affect the average rating or review count.
