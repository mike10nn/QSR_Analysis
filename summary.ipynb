```python
# Summary statistics for numeric socioeconomic variables

socioeconomic.describe()
```

From this result we can understand:
<br>1, How large Italian regions are in population. Italian regions are very different in population size. The average region has 2.95 million people but the median has only 1.69 million population size. Meaning, very few large regions are pulling the average up. Therefore, Most Italian regions are relatively small or medium-sized, while a few large regions contain a very high share of the national population.

2, How different are regions is population density. they vary very much in how crowded they are. some regions have low density, around 37.6 people per km2 while others have 442 people per km2. Therefore, QSR outlets may be more concentrated in dense regions because dense areas have more foot traffic, more urban centers, and more commercial activity.

3, There is also a clear income difference among the regions. Higher-income regions may support more QSR outlets, especially brands with higher price ranges, but this needs to be tested later with correlation and regression.

4, The maximum tourism flow is huge compared to the minimum. Some regions receive less than 500,000 tourist flows, while the highest region receives about 71.9 million. Telling us that, tourism is concentrated in small number of regions

5, Median income also varies strongly across regions. Some regions have much stronger household income conditions than others, which may influence restaurant demand, brand location strategy, and price-range distribution.
