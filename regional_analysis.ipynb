```python
import pandas as pd
```

```python
outlets = pd.read_csv("Data/QSR_outlet_dataset_cleaned.csv", keep_default_na=False)
socioeconomic = pd.read_csv("Data/QSR_socioeconomic_dataset_cleaned.csv")
```

```python
outlets[[
    "Region",
    "Province",
    "City",
    "Price_Range_Cleaned",
    "Customer_Rating",
    "Review_Count",
    "Latitude",
    "Longitude"
]].isna().sum()
```

```python
regional_qsr = outlets.groupby("Region").agg(
    Outlet_Count=("Outlet ID", "count"),
    Average_Rating=("Customer_Rating", "mean"),
    Average_Review_Count=("Review_Count", "mean")
).reset_index()

regional_qsr
```

```python
regional_analysis = regional_qsr.merge(
    socioeconomic,
    on="Region",
    how="left"
)

regional_analysis.head()
```

```python
regional_analysis["Outlets_per_100k_people"] = (
    regional_analysis["Outlet_Count"] / regional_analysis["Population"] * 100000
)

regional_analysis.head()
```

```python
regional_analysis = regional_analysis[[
    "Region",
    "Outlet_Count",
    "Population",
    "Outlets_per_100k_people",
    "Average_Rating",
    "Average_Review_Count",
    "Disposable Income per capita",
    "Total Tourism Flow",
    "Population Density",
    "Median Income"
]]

regional_analysis.head()
```

```python
regional_analysis.sort_values(
    by = "Outlet_Count",
    ascending=False
)[["Region", "Outlet_Count"]]
```

### Regional Ranking by Total Outlet Count

Lombardia has the highest number of outlets, followed by Lazio, Piemonte, Veneto, and Emilia-Romagna.

This result suggests that QSR outlets are most concentrated in larger and economically developed regions, especially those containing major urban areas such as Milano, Roma, Torino, Venezia, and Bologna. However, this ranking only shows the absolute number of outlets and does not account for differences in regional population size. So we move on to the next piece of code:

```python
regional_analysis.sort_values(
    "Outlets_per_100k_people",
    ascending=False
)[["Region", "Outlets_per_100k_people"]]
```

### Regional Ranking by Outlets Per 100,000 People

This result shows the relative concentration of QSR outlets after normalizing for regional population size. Lombardia and Lazio are still the two highest-ranking regions, meaning that their high outlet numbers are not only due to large populations but also reflect a relatively strong QSR presence per resident.

Piemonte, Emilia-Romagna, and Veneto also remain among the highest-ranked regions. which suggests that QSR outlet concentration is strongest in several northern and central regions with large urban centers, strong commercial activity, and important transport or tourism flows.

smaller regions, such as Valle d'Aosta, have very low outlet counts in absolute terms but a higher per-capita value because their population is small.

```python
per_capita_comparison = regional_analysis[[
    "Region",
    "Outlet_Count",
    "Population",
    "Outlets_per_100k_people"
]].sort_values(
    "Outlets_per_100k_people",
    ascending=False
)

per_capita_comparison
```

### All together

This table shows the original outlet count and regional populatio and Adding population makes it easier to understand why some regions move differently in the normalized ranking.

### Summary of Regional Analysis

The regional analysis shows that QSR outlets are unevenly distributed across Italian regions. Lombardia and Lazio have the highest number of outlets in absolute terms, and they also remain the highest after normalizing by population. Which means that these regions have both strong absolute QSR presence and high outlet concentration relative to their populations.

The per-capita or the normalized indicator allows regions with different population sizes to be compared more fairly. Some smaller regions, such as Valle d'Aosta, have low outlet counts but relatively high outlets per 100k people because of their smaller population base. Meanwhile, some larger southern regions have lower per-capita values, suggesting a weaker relative concentration of QSR outlets.

Overall, the results suggest that QSR distribution is concentrated in larger, economically important regions, especially those with major urban centers and commercial activity.
