```python
import pandas as pd
```

```python
outlets = pd.read_csv("Data/QSR_outlet_dataset_cleaned.csv", keep_default_na=False)
socioeconomic = pd.read_csv("Data/QSR_socioeconomic_dataset_cleaned.csv")
```

```python
outlets[[
    "Region",
    "Province",
    "City",
    "Price_Range_Cleaned",
    "Customer_Rating",
    "Review_Count",
    "Latitude",
    "Longitude"
]].isna().sum()
```

```python
regional_qsr = outlets.groupby("Region").agg(
    Outlet_Count=("Outlet ID", "count"),
    Average_Rating=("Customer_Rating", "mean"),
    Average_Review_Count=("Review_Count", "mean")
).reset_index()

regional_qsr
```

```python
regional_analysis = regional_qsr.merge(
    socioeconomic,
    on="Region",
    how="left"
)

regional_analysis.head()
```

```python
regional_analysis["Outlets_per_100k_people"] = (
    regional_analysis["Outlet_Count"] / regional_analysis["Population"] * 100000
)

regional_analysis.head()
```

```python
regional_analysis = regional_analysis[[
    "Region",
    "Outlet_Count",
    "Population",
    "Outlets_per_100k_people",
    "Average_Rating",
    "Average_Review_Count",
    "Disposable Income per capita",
    "Total Tourism Flow",
    "Population Density",
    "Median Income"
]]

regional_analysis
```

```python
regional_analysis.sort_values(
    "Outlet_Count",
    ascending=False
)[["Region", "Outlet_Count"]]
```

### Regional Ranking by Total Outlet Count

The ranking by total outlet count shows the absolute distribution of QSR outlets across Italian regions. Lombardia has the highest number of outlets, followed by Lazio, Piemonte, Veneto, and Emilia-Romagna.

This result suggests that QSR outlets are most concentrated in larger and economically important regions, especially those containing major urban areas such as Milano, Roma, Torino, Venezia, and Bologna. However, this ranking only shows the absolute number of outlets and does not account for differences in regional population size. For this reason, the next step is to compare these results with outlets per 100,000 people.

```python
regional_analysis.sort_values(
    "Outlets_per_100k_people",
    ascending=False
)[["Region", "Outlets_per_100k_people"]]
```

### Regional Ranking by Outlets Per 100,000 People

The outlets per 100,000 people ranking shows the relative concentration of QSR outlets after adjusting for regional population size. Lombardia and Lazio remain the two highest-ranking regions, meaning that their high outlet numbers are not only due to large populations but also reflect a relatively strong QSR presence per resident.

Piemonte, Emilia-Romagna, and Veneto also remain among the highest-ranked regions. This suggests that QSR outlet concentration is strongest in several northern and central regions with large urban centers, strong commercial activity, and important transport or tourism flows.

Compared with the raw outlet count ranking, the per-capita ranking provides a fairer comparison across regions because it accounts for population size.

```python
ranking_comparison = regional_analysis[[
    "Region",
    "Outlet_Count",
    "Outlets_per_100k_people"
]].sort_values(
    "Outlet_Count",
    ascending=False
)

ranking_comparison
```

### Comparison of Raw and Per-Capita Outlet Distribution

The comparison table shows both the absolute number of QSR outlets and the normalized outlet concentration per 100,000 people. Lombardia and Lazio rank highest in both measures, indicating that they have both a large number of outlets and a high outlet concentration relative to population.

smaller regions, such as Valle d'Aosta, have very low outlet counts in absolute terms but a higher per-capita value because their population is small.

so the comparison suggests that QSR distribution is partly related to population size, but the per-capita measure gives a clearer view of relative regional concentration.

```python
per_capita_comparison = regional_analysis[[
    "Region",
    "Outlet_Count",
    "Population",
    "Outlets_per_100k_people"
]].sort_values(
    "Outlets_per_100k_people",
    ascending=False
)

per_capita_comparison
```

### Per-Capita Ranking With Population Context

This table ranks regions by QSR outlets per 100,000 people while also showing the original outlet count and regional population. Adding population makes it easier to understand why some regions move differently in the normalized ranking.

Lombardia and Lazio remain at the top because they have both large outlet counts and high outlet concentration relative to their populations. Valle d'Aosta has only 5 outlets, but its small population gives it a relatively high per-capita value. In contrast, regions such as Campania, Sicilia, and Puglia have larger populations but lower outlets per 100,000 people, which suggests a lower relative QSR concentration.

### Summary of Regional Analysis

The regional analysis shows that QSR outlets are unevenly distributed across Italian regions. Lombardia and Lazio have the highest number of outlets in absolute terms, and they also remain the highest after normalizing by population. This suggests that these regions have both strong absolute QSR presence and high outlet concentration relative to their populations.

The per-capita indicator is important because it allows regions with different population sizes to be compared more fairly. Some smaller regions, such as Valle d'Aosta, have low outlet counts but relatively high outlets per 100,000 people because of their smaller population base. Meanwhile, some larger southern regions have lower per-capita values, suggesting a weaker relative concentration of QSR outlets.

Overall, the results suggest that QSR distribution is concentrated in larger, economically important regions, especially those with major urban centers and commercial activity. This regional table will provide the basis for later spatial mapping, correlation analysis, and regression analysis.
